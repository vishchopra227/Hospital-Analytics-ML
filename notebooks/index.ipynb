```python
from scipy import stats

ratings = df["hospital_overall_rating"].dropna()

print("Hospital Rating Statistics")
print("--------------------------")
print("Mean:", round(ratings.mean(), 2))
print("Median:", ratings.median())
print("Standard Deviation:", round(ratings.std(), 2))
print("Minimum:", ratings.min())
print("Maximum:", ratings.max())
```

```python
ownership_data = df[
    ["hospital_ownership", "hospital_overall_rating"]
].dropna()

ownership_groups = [
    group["hospital_overall_rating"].values
    for _, group in ownership_data.groupby("hospital_ownership")
    if len(group) >= 5
]

stat, p = stats.kruskal(*ownership_groups)

print("Kruskal-Wallis Test: Hospital Ownership")
print("Statistic:", round(stat, 4))
print("P-value:", round(p, 6))

if p < 0.05:
    print("Result: Significant difference in ratings across ownership groups.")
else:
    print("Result: No statistically significant difference detected.")
```

```python
emergency_data = df[
    ["emergency_services", "hospital_overall_rating"]
].dropna()

emergency_data["emergency_services"] = (
    emergency_data["emergency_services"]
    .str.strip()
    .str.lower()
)

yes_ratings = emergency_data.loc[
    emergency_data["emergency_services"] == "yes",
    "hospital_overall_rating"
]

no_ratings = emergency_data.loc[
    emergency_data["emergency_services"] == "no",
    "hospital_overall_rating"
]

stat, p = stats.mannwhitneyu(
    yes_ratings,
    no_ratings,
    alternative="two-sided"
)

print("Mann-Whitney U Test: Emergency Services")
print("Statistic:", round(stat, 4))
print("P-value:", round(p, 6))

print("\nMedian rating (Emergency Yes):", yes_ratings.median())
print("Median rating (Emergency No):", no_ratings.median())

if p < 0.05:
    print("Result: Significant difference in rating distributions.")
else:
    print("Result: No statistically significant difference detected.")
```

```python
state_data = df[
    ["state", "hospital_overall_rating"]
].dropna()

state_groups = [
    group["hospital_overall_rating"].values
    for _, group in state_data.groupby("state")
    if len(group) >= 10
]

stat, p = stats.kruskal(*state_groups)

print("Kruskal-Wallis Test: State-wise Ratings")
print("States included:", len(state_groups))
print("Statistic:", round(stat, 4))
print("P-value:", round(p, 6))

if p < 0.05:
    print("Result: Significant differences in ratings across states.")
else:
    print("Result: No statistically significant differences detected.")
```

```python
results = []

tests = [
    ("Hospital Ownership", ownership_groups, "kruskal"),
    ("Emergency Services", [yes_ratings, no_ratings], "mannwhitney"),
    ("State-wise Ratings", state_groups, "kruskal")
]

for name, groups, test in tests:
    if len(groups) < 2 or any(len(g) == 0 for g in groups):
        results.append([name, np.nan])
        continue

    if test == "kruskal":
        p = stats.kruskal(*groups).pvalue
    else:
        p = stats.mannwhitneyu(
            groups[0],
            groups[1],
            alternative="two-sided"
        ).pvalue

    results.append([name, p])

summary = pd.DataFrame(
    results,
    columns=["Statistical Test", "P-value"]
)

summary["Adjusted P-value"] = (
    summary["P-value"] * 3
).clip(upper=1)

summary["Conclusion"] = np.where(
    summary["Adjusted P-value"] < 0.05,
    "Significant",
    "Not Significant"
)

display(summary.round(6))
```

```python
target = "hospital_overall_rating"

feature_columns = [
    "state",
    "hospital_type",
    "hospital_ownership",
    "emergency_services"
]

features = [
    col for col in feature_columns
    if col in df.columns
]

df[target] = pd.to_numeric(
    df[target],
    errors="coerce"
)

ml_df = df[
    df[target].isin([1, 2, 3, 4, 5])
].copy()

X = ml_df[features]
y = ml_df[target].astype(int)

print("Features used:", features)
print("Dataset shape:", X.shape)

print("\nTarget distribution:")
print(y.value_counts().sort_index())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

stratify_target = (
    y if y.value_counts().min() >= 2 else None
)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=stratify_target
)

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(
        strategy="constant",
        fill_value="Missing"
    )),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore"
    ))
])

preprocessor = ColumnTransformer([
    ("categorical", categorical_transformer, features)
])

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.dummy import DummyClassifier

model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        class_weight="balanced",
        min_samples_leaf=5,
        n_jobs=-1
    ))
])

baseline = DummyClassifier(
    strategy="most_frequent"
)

baseline.fit(X_train, y_train)
model.fit(X_train, y_train)

baseline_pred = baseline.predict(X_test)
y_pred = model.predict(X_test)

print("Model training completed successfully.")
```

```python
from sklearn.metrics import accuracy_score, f1_score

results = pd.DataFrame({
    "Model": ["Baseline", "Random Forest"],
    "Accuracy": [
        accuracy_score(y_test, baseline_pred),
        accuracy_score(y_test, y_pred)
    ],
    "Macro F1": [
        f1_score(
            y_test,
            baseline_pred,
            average="macro",
            zero_division=0
        ),
        f1_score(
            y_test,
            y_pred,
            average="macro",
            zero_division=0
        )
    ]
})

display(results.round(3))
```

```python
from sklearn.metrics import classification_report, ConfusionMatrixDisplay

print("Random Forest Classification Report:")
print(classification_report(
    y_test,
    y_pred,
    labels=[1, 2, 3, 4, 5],
    zero_division=0
))

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    labels=[1, 2, 3, 4, 5],
    cmap="Blues",
    values_format="d"
)

plt.title("Hospital Rating Prediction — Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
# Extract feature importance
encoder = (
    model.named_steps["preprocessor"]
    .named_transformers_["categorical"]
    .named_steps["encoder"]
)

feature_names = encoder.get_feature_names_out(features)
importances = model.named_steps["classifier"].feature_importances_

feature_importance = (
    pd.DataFrame({
        "Feature": feature_names,
        "Importance": importances
    })
    .sort_values("Importance", ascending=False)
)

display(feature_importance.head(15))
```

```python
top_features = feature_importance.head(15).sort_values("Importance")

plt.figure(figsize=(10, 6))
plt.barh(
    top_features["Feature"],
    top_features["Importance"]
)

plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("Top 15 Feature Importances — Random Forest")
plt.tight_layout()
plt.show()
```

```python
# Get encoded feature names
encoder = (
    model.named_steps["preprocessor"]
    .named_transformers_["categorical"]
    .named_steps["encoder"]
)

encoded_features = encoder.get_feature_names_out(features)

# Get importance of each encoded feature
importances = pd.Series(
    model.named_steps["classifier"].feature_importances_,
    index=encoded_features
)

# Aggregate encoded categories into original features
feature_importance = pd.Series({
    feature: importances[
        [name.startswith(feature + "_")
         for name in importances.index]
    ].sum()
    for feature in features
}).sort_values(ascending=False)

display(
    feature_importance
    .to_frame("Importance")
    .round(4)
)
```

```python
plt.figure(figsize=(8, 4))

feature_importance.sort_values().plot(kind="barh")

plt.title("Hospital Characteristics — Feature Importance")
plt.xlabel("Random Forest Feature Importance")
plt.tight_layout()
plt.show()
```
